## Video 10 y 11: Operaciones Matemáticas y Estadísticas
* **Análisis visual:** Numpy permite hacer operaciones aritméticas directas entre arrays completos sin usar bucles for. Por ejemplo, al sumar dos arrays, se suman sus elementos posición por posición. Además, incluye funciones estadísticas potentes como `.max()` para encontrar el valor máximo y `.argmax()` para saber exactamente en qué posición (índice) se encuentra ese valor.

## Video 12: Copia de Arrays (.copy)
* **Análisis visual:** Un concepto crucial en Numpy: si igualas una variable a un array existente (`b = a`), ambas apuntarán al mismo espacio de memoria; si cambias una, se altera la otra. Para crear un respaldo real e independiente sin afectar el original, se debe usar obligatoriamente la función `.copy()`.

```python
import numpy as np

# ------------------------------------------------------------------------------
# 1. OPERACIONES BÁSICAS Y ESTADÍSTICA (Videos 10 y 11)
# ------------------------------------------------------------------------------
array_A = np.array([10, 20, 30])
array_B = np.array([1, 2, 3])

# Lógica: Suma directa elemento por elemento
suma_arrays = array_A + array_B
print("Suma directa de arreglos:", suma_arrays)

# Lógica: Funciones estadísticas (.max y .argmax)
edades = np.array([25, 42, 18, 65, 30])
print("\nLa edad máxima es:", edades.max())
print("El índice donde está esa edad máxima es:", edades.argmax())

# ------------------------------------------------------------------------------
# 2. COPIA INDEPENDIENTE DE ARRAYS (Video 12)
# ------------------------------------------------------------------------------
array_original = np.array([5, 5, 5])

# Lógica: Usamos .copy() para crear un clon seguro en otro espacio de memoria
respaldo_seguro = array_original.copy()

# Modificamos el respaldo
respaldo_seguro[0] = 99

print("\nArray original (intacto):", array_original)
print("Respaldo modificado:", respaldo_seguro)
```
